### Multi-level Data Exploration Overview

Multi-level data exploration refers to the process of examining and analyzing data at different levels of detail or abstraction to uncover patterns, insights, and relationships. It’s like zooming in and out of your data to see both the big picture and the finer details from high-level to low-level exploration.

#### High-Level Exploration – Look at broad summaries:
- Mean, median, min, max
- istograms, box plots
- Count of categories, missing data

#### Mid-Level Exploration – Compare groups or variables:
- Correlations between variables
- Group-by aggregations (e.g., average sales per region)
- Visual comparisons like bar plots or scatter plots

#### Low-Level Exploration – Dive deep into specific cases:
- Investigating outliers
- Looking at individual records
- Tracing data transformations

It’s often used in exploration data analysis (EDA), business intelligence dashboards, or machine learning prep, helping analysts or data scientists move between summaries and specifics to better understand the data.

## Check for missing values

```python
print("Checking for missing values...")

try:
    # Count missing values in each column
    missing_values = data.isnull().sum()

    # Display columns containing missing values
    missing_columns = missing_values[missing_values > 0]

    print("\nColumns with missing values:")
    print(missing_columns)

    # Display total missing values
    print("\nTotal missing values:", missing_values.sum())

    # Display number of affected columns
    print("Columns with missing data:", len(missing_columns))

except (ValueError, TypeError) as e:
    print("Error checking missing values:", e)
    raise
```

## Handle missing values

```python
print("Handling missing values...")

try:
    # Create a copy to preserve the original dataset
    cleaned_data = data.copy()

    # Convert the date column to datetime
    cleaned_data["date"] = pd.to_datetime(
        cleaned_data["date"],
        errors="raise"
    )

    # Check for missing values and remove incomplete rows if needed
    if cleaned_data.isnull().sum().sum() > 0:
        cleaned_data = cleaned_data.dropna()
        print("Rows containing missing values were removed.")
    else:
        print("No missing values found. No rows removed.")

    # Verify the cleaned dataset
    print("Cleaned dataset shape:", cleaned_data.shape)
    print("Remaining missing values:", cleaned_data.isnull().sum().sum())
    print("Date column type:", cleaned_data["date"].dtype)

except (ValueError, TypeError, KeyError) as e:
    print("Error handling missing values:", e)
    raise
```

## Create Visualizations

### Visualization 1: Correlation Heatmap Between Countries (Top 10 by Total Deaths)

```python
AGGREGATE_COLUMNS = {
    "World", "Africa", "Asia", "Europe", "North America", "South America",
    "Oceania", "European Union (27)", "High-income countries",
    "Upper-middle-income countries", "Lower-middle-income countries",
    "Low-income countries", "European Union", "International"
}
def country_columns(frame):
    return [col for col in frame.select_dtypes(include=[np.number]).columns
            if col not in AGGREGATE_COLUMNS and not col.endswith(" countries")]

def create_correlation_heatmap(data):
    print("Aggregating cumulative deaths by country (latest date)...")
    countries = country_columns(data)
    latest_totals = data.sort_values("date")[countries].iloc[-1]
    top_10 = latest_totals.nlargest(10).index.tolist()
    print("Top 10 countries:", top_10)
    country_data = data.sort_values("date")[top_10]
    correlation_matrix = country_data.corr()
    print("Correlation matrix shape:", correlation_matrix.shape)
    os.makedirs("output", exist_ok=True)
    plt.figure(figsize=(11, 9))
    sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm",
                vmin=-1, vmax=1, square=True, linewidths=0.5)
    plt.title("Correlation of Normalized Cumulative Deaths: Top 10 Countries")
    plt.xlabel("Country")
    plt.ylabel("Country")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.savefig("output/correlation_heatmap.png", dpi=180, bbox_inches="tight")
    plt.close()
    display(Image(filename="output/correlation_heatmap.png"))
```

### Visualization 2: Hierarchical Bar Chart (Top 20 Countries by Total Deaths)

```python
def create_bar_chart(data):
    print("Creating bar chart for top 20 countries...")
    latest = data.sort_values("date")[country_columns(data)].iloc[-1]
    top_20 = latest.nlargest(20).sort_values()
    os.makedirs("output", exist_ok=True)
    plt.figure(figsize=(12, 9))
    bars = plt.barh(top_20.index, top_20.values, color=plt.cm.viridis(np.linspace(0.2, 0.9, len(top_20))))
    plt.title("Top 20 Countries by Reported Cumulative COVID-19 Deaths")
    plt.xlabel("Cumulative Reported Deaths")
    plt.ylabel("Country")
    plt.grid(axis="x", alpha=0.25)
    plt.tight_layout()
    plt.savefig("output/hierarchical_bar_chart.png", dpi=180, bbox_inches="tight")
    plt.close()
    display(Image(filename="output/hierarchical_bar_chart.png"))
```

### Visualization 3: Time Series Chart (Global Trends in Total Deaths)

```python
def create_time_series_chart(data):
    print("Creating global time series chart...")
    ordered = data.sort_values("date")
    # The World column is already a global total; summing all regions would double count.
    global_total = ordered["World"] if "World" in ordered else ordered[country_columns(ordered)].sum(axis=1)
    os.makedirs("output", exist_ok=True)
    plt.figure(figsize=(13, 6))
    plt.plot(ordered["date"], global_total, color="navy", linewidth=2)
    peak_idx = global_total.idxmax()
    peak_date = ordered.loc[peak_idx, "date"]
    peak_value = global_total.loc[peak_idx]
    plt.annotate(f"Latest peak: {peak_value:.3f}", xy=(peak_date, peak_value),
                 xytext=(-125, -45), textcoords="offset points",
                 arrowprops={"arrowstyle": "->"})
    plt.title("Global Normalized Cumulative COVID-19 Deaths Over Time")
    plt.xlabel("Date")
    plt.ylabel("Normalized Cumulative Deaths (World)")
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig("output/annotated_time_series.png", dpi=180, bbox_inches="tight")
    plt.close()
    display(Image(filename="output/annotated_time_series.png"))
```

## Load the Dataset

```python
original_animation_data = pd.read_csv("data/COVID_total_deaths.csv")
print("Original dataset loaded for animation:", original_animation_data.shape)
```

## Clean and Preprocess the Data

```python
def clean_data(data):
    print("\nCleaning the dataset...")
    cleaned = data.copy()
    cleaned["date"] = pd.to_datetime(cleaned["date"], errors="raise")
    numeric = cleaned.select_dtypes(include=[np.number]).columns
    cleaned = cleaned.loc[:, ["date"] + [c for c in numeric if cleaned[c].fillna(0).sum() > 0]]
    active_columns = cleaned.columns.drop("date")
    cleaned = cleaned.loc[cleaned[active_columns].fillna(0).sum(axis=1) > 0].copy()
    cleaned = cleaned.sort_values("date")
    print("Filtered dataset shape:", cleaned.shape)
    return cleaned
```

```python
cleaned_animation_data = clean_data(original_animation_data)
```

## Select Top 10 Countries

```python
def select_top_countries(data):
    print("\nSelecting top 10 countries by total deaths...")
    country_names = country_columns(data)
    # Use most recent cumulative totals, not a sum of daily cumulative counts.
    most_recent = data.sort_values("date")[country_names].iloc[-1]
    top10 = most_recent.nlargest(10).index.tolist()
    print("\nTop 10 individual countries by total deaths:")
    print(most_recent.loc[top10].to_string())
    filtered = data[["date"] + top10].copy().set_index("date").sort_index()
    weekly = filtered.resample("W").last().ffill().reset_index()
    print(f"\nFiltered dataset shape after selecting and weekly sampling: {weekly.shape}")
    return weekly
```

```python
top_countries_data = select_top_countries(cleaned_animation_data)
```

## Prepare Data for Animation

```python
print("Preparing data for animation...")
animated_long = top_countries_data.melt(id_vars="date", var_name="Country", value_name="Deaths")
animated_long = animated_long.sort_values(["date", "Country"])
```

```python
print("Data prepared successfully!")
display(animated_long.head())
print("Animation frames:", animated_long["date"].nunique())
```

## Create the Matplotlib Animation

```python
print("Creating the animation...ignore static output")
fig, ax = plt.subplots(figsize=(12, 7))
country_names_animation = top_countries_data.columns.drop("date").tolist()
lines = {name: ax.plot([], [], label=name, linewidth=1.7)[0] for name in country_names_animation}
frame_dates = top_countries_data["date"].tolist()
ax.set_xlim(frame_dates[0], frame_dates[-1])
ax.set_ylim(0, float(top_countries_data[country_names_animation].max().max()) * 1.08)
ax.set_xlabel("Date")
ax.set_ylabel("Cumulative COVID-19 Deaths")
ax.set_title("Top 10 Countries: COVID-19 Deaths Over Time")
ax.legend(loc="upper left", fontsize=8, ncol=2)
ax.grid(alpha=0.25)
plt.tight_layout()
plt.close(fig)  # Avoid an extra static chart in the notebook
```

### Function to initialize the animation

```python
def init():
    for line in lines.values():
        line.set_data([], [])
    return tuple(lines.values())
```

### Function to update the animation frame-by-frame

```python
def update(frame):
    current = top_countries_data.iloc[:frame + 1]
    for name, line in lines.items():
        line.set_data(current["date"], current[name])
    ax.set_title(f"Top 10 Countries: Cumulative Deaths — {current['date'].iloc[-1]:%Y-%m-%d}")
    return tuple(lines.values())
```

### Create the animation

<div class="alert alert-block alert-info">
    <span style="color:black"><b><i>Hints:</i></b> Use <b><mark>FuncAnimation()</mark></b> to initate the animation visualization. Use <b><mark>HTML({variable}.to_jshtml())</mark></b> to render your animation inside the notebook in an html exportable format.</span>
</div>

```python

from IPython.display import HTML

# Limit the number of embedded frames to avoid size limits
frame_indices = np.unique(
    np.linspace(0, len(top_countries_data) - 1, 80).astype(int)
)

animation = FuncAnimation(
    fig,
    update,
    frames=frame_indices,
    init_func=init,
    interval=200,
    blit=False,
    repeat=False
)

display(HTML(animation.to_jshtml()))
print("Animation rendered successfully.")
```
