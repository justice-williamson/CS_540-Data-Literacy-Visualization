### Data File 1 dataframe
Here we will load the data_file_1 example data, then pivot and clean it in preparation for merging with the other two selected sample files.

For your submission you will adjust the code to reflect your selected dataset.


```python
# Load the first dataset and inspect its structure.
data1 = pd.read_csv(data_file_1)

print("NYC Weather 2018")
print("Dataset dimensions:", data1.shape)
print("\nFirst five rows:")
display(data1.head())

print("\nColumn information:")
data1.info()

print("\nMissing values:")
print(data1.isnull().sum())

print("\nDuplicate rows:", data1.duplicated().sum())
```

```python
# Clean the NYC Weather 2018 dataset.
# Keep the original dataframe unchanged.

data1_clean = data1.copy()

# Standardize column names.
data1_clean.columns = data1_clean.columns.str.strip().str.lower()

# Convert date values to datetime format.
data1_clean["date"] = pd.to_datetime(
    data1_clean["date"], errors="coerce"
)

# Convert weather measurements into numeric values.
data1_clean["value"] = pd.to_numeric(
    data1_clean["value"], errors="coerce"
)

# Remove rows missing essential information.
data1_clean = data1_clean.dropna(
    subset=["date", "station", "datatype", "value"]
)

# Remove duplicate station/date/measurement combinations.
data1_clean = data1_clean.drop_duplicates(
    subset=["date", "station", "datatype"]
)

# Display cleaning results.
print("NYC Weather 2018 - Cleaned Dataset")
print("Dataset dimensions:", data1_clean.shape)

print("\nMissing values:")
print(data1_clean.isnull().sum())

print("\nDuplicate observations:",
      data1_clean.duplicated(
          subset=["date", "station", "datatype"]
      ).sum())

print("\nWeather measurement types:")
print(data1_clean["datatype"].value_counts())

display(data1_clean.head())
```

```python
# Reshape the cleaned weather data so each station and date
# has one row with separate columns for each measurement.

data1_wide = data1_clean.pivot_table(
    index=["date", "station"],
    columns="datatype",
    values="value",
    aggfunc="first"
).reset_index()

# Remove the column-axis label for a cleaner table.
data1_wide.columns.name = None

print("Weather dataset after reshaping:")
print("Dimensions:", data1_wide.shape)

print("\nFirst five rows:")
display(data1_wide.head())

print("\nMissing values by column:")
print(data1_wide.isnull().sum())
```

### Data File 2 dataframe
Here we will load the data_file_2 example data, then pivot and clean it in preparation for merging with the other two selected sample files.

For your submission you will adjust the code to reflect your selected dataset.

```python
# Load the NYC temperature dataset and inspect its structure.
# This helps identify data types, missing values, and duplicates
# before cleaning and merging.

data2 = pd.read_csv(data_file_2)

print("NYC Temperatures")
print("Dataset dimensions:", data2.shape)

print("\nFirst five rows:")
display(data2.head())

print("\nColumn information:")
data2.info()

print("\nMissing values:")
print(data2.isnull().sum())

print("\nDuplicate rows:", data2.duplicated().sum())
```

```python
# Clean the NYC temperature dataset before merging.
data2_clean = data2.copy()

# Standardize column names for consistency.
data2_clean.columns = data2_clean.columns.str.strip().str.lower()

# Convert dates and values into appropriate data types.
data2_clean["date"] = pd.to_datetime(data2_clean["date"])
data2_clean["value"] = pd.to_numeric(
    data2_clean["value"], errors="coerce"
)

# Remove duplicate records based on station, date, and datatype.
data2_clean = data2_clean.drop_duplicates(
    subset=["date", "station", "datatype"]
)

# Check for missing values after conversion.
print("Missing values:")
print(data2_clean.isnull().sum())

# Display cleaned data and dimensions.
print("\nCleaned dataset dimensions:", data2_clean.shape)
display(data2_clean.head())
```

```python
# Reshape the cleaned NYC temperature dataset.
# Each row represents one station on one date.

data2_wide = data2_clean.pivot_table(
    index=["date", "station"],
    columns="datatype",
    values="value",
    aggfunc="first"
).reset_index()

# Remove the extra column-axis label.
data2_wide.columns.name = None

# Rename temperature columns for clarity.
data2_wide = data2_wide.rename(columns={
    "TAVG": "temp_avg",
    "TMAX": "temp_max",
    "TMIN": "temp_min"
})

# Verify the reshaped temperature dataset.
print("Temperature dataset dimensions:", data2_wide.shape)
display(data2_wide.head())
```

### Data Merging
Here is the example of a merge of the three sample files. You will need to adjust these blocks for how you wish to merge your selected data files.

<div class="alert alert-block alert-danger">
    <span style="color:black"><b><i>Example Data:</i></b> Keep in mind that the output below are from the sample data and will need replaced before running the block with your dataset for how you need merge your dataset.</span>
</div>

```python
# Merge the cleaned NYC temperature data with additional
# weather observations using matching dates and stations.

# Remove overlapping temperature columns from the weather data.
weather_extra = data1_wide.drop(
    columns=["TAVG", "TMAX", "TMIN"],
    errors="ignore"
)

# Use a left join to preserve the October temperature records.
final_merged_data = pd.merge(
    data2_wide,
    weather_extra,
    on=["date", "station"],
    how="left",
    validate="one_to_one",
    indicator=True
)

# Inspect the merge results.
print("Merged dataset dimensions:", final_merged_data.shape)

print("\nMerge validation:")
print(final_merged_data["_merge"].value_counts())

print("\nMissing values:")
print(final_merged_data.isnull().sum())

print("\nDuplicate date/station records:",
      final_merged_data.duplicated(
          subset=["date", "station"]
      ).sum())

display(final_merged_data.head())
```

```python
# Final validation of the NYC weather and temperature dataset.
# Remove the merge indicator before exporting.

final_merged_data = final_merged_data.drop(
    columns=["_merge"],
    errors="ignore"
)

# Check the final dataset dimensions and column types.
print("Final dataset dimensions:", final_merged_data.shape)

print("\nColumn information:")
final_merged_data.info()

# Check for duplicate date/station records.
print("\nDuplicate records:",
      final_merged_data.duplicated(
          subset=["date", "station"]
      ).sum())

# Check missing values without replacing them with zero.
print("\nMissing values:")
print(final_merged_data.isnull().sum())

# Verify that all expected October dates are included.
print("\nUnique dates:", final_merged_data["date"].nunique())
print("Unique stations:", final_merged_data["station"].nunique())

# Display a sample of the finished dataset.
display(final_merged_data.head())
```
